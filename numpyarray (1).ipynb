```python
def convert_sqft_to_float(x):
     try:
        if'-'in x:
            tokens = x.split('-')
            return (float(tokens[0]) + float(tokens[1])) / 2
        return float(x)
     except:
            return np.nan  #if conversion fails, return nan
```

```python
de['total_sqft_cleaned'] = de['total_sqft'].apply(convert_sqft_to_float)
```

```python
from scipy import stats
```

```python
#Sample data: heights of students (in cm)

sample_heights = [158, 162, 161, 159, 160, 157, 163, 160]

# Null hypothesis: mean = 160 
t_stat, p_value = stats.ttest_1samp(sample_heights, popmean=160)

print("t-statistic:", round(t_stat, 3))
print("p-value:", round(p_value, 3))

# Final decision

if p_value < 0.05: 
    print("Reject Null Hypothesis (There is a significant difference)")
else:
    print("Fail to Reject Null Hypothesis (No significant difference)")
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

print("Mean:", df.describe().loc['mean'])
print("Median:", df.median(numeric_only=True))
print("Mode:", df.mode(numeric_only=True).iloc[0])

Q1 = df.select_dtypes(include='number').quantile(0.25)
Q3 = df.select_dtypes(include='number').quantile(0.75)
IQR = Q3 - Q1
outliers = ((df.select_dtypes(include='number') < (Q1 - 1.5 * IQR)) |
            (df.select_dtypes(include='number') > (Q3 + 1.5 * IQR)))
print("Outliers:", outliers)

label_encoder = LabelEncoder()
df['Gender_encoded'] = label_encoder.fit_transform(df['Gender'])

numeric_cols = ['Math', 'Physics', 'Chemistry', 'English', 'CS']
scaler = MinMaxScaler()
df_scaled = df.copy()
df_scaled[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("Final Scaled Data:", df_scaled)
```
